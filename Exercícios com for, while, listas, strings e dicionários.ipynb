```python
numros = [3, 7, 2, 8, 10]
soma = 0

for num in numros:
    soma += num

print("A soma dos números é:", soma)
```

```python
palavra = input("Dígite a palavra: ")
contador = 0
for letra in palavra.lower():
    if letra in "aeiou":
        contador += 1
        print(f"A palavra '{palavra}' tem {contador} vogais.")
```

Atividade3

```python
nomes = ["Ana", "João", "Carlos", "Maria"]

invertidos = []

i = len(nomes) - 1

while i >= 0:
    invertidos.append(nomes[i])
    i -= 1

print("Lista original:", nomes)
print("Lista invertida:", invertidos)
```

Actividade4

```python
numeros = [15, 42, 7, 89, 23, 56, 102, 3]

maior = numeros[0]

for num in numeros:
    if num > maior:
        maior = num

print("O maior valor da lista é:", maior)
```

atividade5

```python
valores = [1, 2, 2, 3, 4, 4, 5]

unicos = []

for v in valores:
    if v not in unicos:
        unicos.append(v)

print("Lista original:", valores)
print("Lista com valores únicos:", unicos)
```

atividade6

```python
palavra = input("Digite uma palavra: ")

invertida = ""
for letra in palavra:
    invertida = letra + invertida

if palavra == invertida:
    print("A palavra é um palíndromo!")
else:
    print("A palavra não é um palíndromo.")
```

Atividade9

```python
numeros = [10, 21, 32, 43, 54]

pares = []

for n in numeros:
    if n % 2 == 0:
        pares.append(n)

print("Lista original:", numeros)
print("Lista com pares:", pares)
```

Atividade10

```python
frase = input("Digite uma frase: ")


palavras = frase.split()


contagem = {}


for p in palavras:
    if p in contagem:
        contagem[p] += 1
    else:
        contagem[p] = 1

print("Contagem de palavras:")
for palavra, qtd in contagem.items():
    print(f"{palavra}: {qtd}")
```
