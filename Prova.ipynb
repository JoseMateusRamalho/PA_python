```python
def CadastrarTime(escala):
    time = {}

    i= 1
    while i < escala+1:
        print(f"\nIntegrante {i}:")
        nome = input("Nome: ")
        idade = int(input("sua idade: "))

        if idade < 18 or idade > 35:
            print("voce nao pode participa\n")
        else:
            print("cadastro concluido\n")
            time[nome] = idade
            i+=1
    return time

CadastrarTime(5)
```
